# Task 05

```python
import numpy as np
file_path = "Iris - Iris.csv"

iris_data = np.genfromtxt(
    "/content/Iris - Iris.csv",
    delimiter = ",",
    dtype = None,
    encoding = "utf-8",
    names = True,
)

print("Column Names and Data Types")
print(iris_data.dtype)


print("\nFirst 5 Rows of Data")
print(iris_data [:5])

print("\nAccessing Columns by Header Name")
print("Sepal Lengths (first 5):", iris_data ["SepalLengthCm"][:5])
print("Species Labels (first 5):", iris_data ["Species"][:5])
```

# Task 06

```python
sepal_length = iris_data["SepalLengthCm"].astype(float)

mean_sl = np.mean(sepal_length)
median_sl = np.median(sepal_length)
std_sl = np.std(sepal_length)

print(f"Task 6 - SepalLengthCm Statistics:")
print(f"  Mean:               {mean_sl:.4f} cm")

print(f"  Median:             {median_sl:.4f} cm")

print(f"  Standard Deviation: {std_sl:.4f} cm")

numeric_columns = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm",
]

print("\n--- Task 6 - Full Statistical Summary of Numeric Columns ---")

for col in numeric_columns:
    col_data = iris_data[col].astype(float)

    c_min = np.min(col_data)

    c_max = np.max(col_data)
    c_var = np.var(col_data)

    p25 = np.percentile(col_data, 25)
    p50 = np.percentile(col_data, 50)

    p75 = np.percentile(col_data, 75)


    print(f"Column: {col}")
    print(f"  Min: {c_min:.2f} | Max: {c_max:.2f} | Variance: {c_var:.4f}")
    print(f"  25th Pct: {p25:.2f} | 50th Pct: {p50:.2f} | 75th Pct: {p75:.2f}")
    print("-" * 50)

unique_species = np.unique(iris_data["Species"])

print("\n--- Task 6 - Feature Means per Species ---")

for species in unique_species:

    species_mask = iris_data["Species"] == species

    print(f"Species: {species}")

    for col in numeric_columns:

        col_species_values = iris_data[col][species_mask].astype(float)
        species_col_mean = np.mean(col_species_values)
        print(f"  - Mean {col:<15}: {species_col_mean:.4f}")
```

```python
import pandas as pd

california_csv_path = "/content/california_housing_test - california_housing_test.csv"


df = pd.read_csv(california_csv_path)

print("=== Task 1.1: First 10 Rows ===")

print(df.head(10))

print("\n=== Task 1.2: Column Data Types ===")
print(df.dtypes)


print("\n=== Task 1.3: Numerical Summary Statistics ===")
print(df.describe())
```

# Task 08

```python
df_missing = df.copy()

idx_bedrooms = df_missing.sample(frac=0.05, random_state=0).index


idx_income = df_missing.sample(frac=0.05, random_state=42).index

df_missing.loc[idx_bedrooms, "total_bedrooms"] = np.nan

df_missing.loc[idx_income, "median_income"] = np.nan

print("=== Task 2.1: Missing Values Count per Column ===")


print(df_missing.isnull().sum())

df_dropped = df_missing.dropna()


print(f"\n=== Task 2.2: Row Count After dropna() ===")
print(
    f"Original rows: {df_missing.shape[0]} | Remaining rows: {df_dropped.shape[0]}"
)

df_filled_mean = df_missing.copy()


mean_bedrooms = df_filled_mean["total_bedrooms"].mean()
df_filled_mean["total_bedrooms"] = df_filled_mean["total_bedrooms"].fillna(
    mean_bedrooms
)

mean_income = df_filled_mean["median_income"].mean()
df_filled_mean["median_income"] = df_filled_mean["median_income"].fillna(
    mean_income
)
print("\n=== Task 2.3: Total Null Values After Mean Imputation ===")
print(df_filled_mean.isnull().sum().sum())
df_filled_mode = df_missing.copy()


mode_bedrooms = df_filled_mode["total_bedrooms"].mode()[0]

df_filled_mode["total_bedrooms"] = df_filled_mode["total_bedrooms"].fillna(
    mode_bedrooms
)
```

# Task 09

```python
income_house_corr = df["median_income"].corr(df["median_house_value"])


print(
    f"=== Task 3.1: Correlation (median_income vs median_house_value) ===\n{income_house_corr:.4f}"
)


df["income_usd"] = df["median_income"] * 10000


df["price_to_income_ratio"] = df["median_house_value"] / df["income_usd"]
df["lat_round"] = df["latitude"].round(1)
df["lon_round"] = df["longitude"].round(1)
area_analysis = (
    df.groupby(["lat_round", "lon_round"])
    .agg(
        avg_house_val=("median_house_value", "mean"),
        avg_income_usd=("income_usd", "mean"),
        avg_price_income_ratio=(
            "price_to_income_ratio",
            "mean",
        ),
        sample_count=("median_house_value", "count"),
    )
    .reset_index()
)

reliable_areas = area_analysis[area_analysis["sample_count"] >= 5]


undervalued_areas = reliable_areas.sort_values(
    by="avg_price_income_ratio", ascending=True
)

print(
    "\n=== Task 3.2: Top 5 Undervalued Areas (Lowest House Value Relative to Income) ==="
)
print(undervalued_areas.head(5))


high_income_areas = reliable_areas.sort_values(
    by="avg_income_usd", ascending=False
)

print(
    "\n=== Task 3.3: Top 5 High-Income Growth Hubs (Highest Regional Income) ==="
)
print(high_income_areas.head(5))
```
